```python
from transformers import pipeline

classifier = pipeline(
    "zero-shot-classification",
    model="facebook/bart-large-mnli"
)

labels = [
    "water leak",
    "energy efficiency",
    "waste recycling",
    "sustainable procurement",
    "accessibility inclusion",
    "governance compliance",
    "other"
]
```

```python
message = message = message = "I am concerned that one of our suppliers may not meet our sustainability policy."

result = classifier(message, labels)

for label, score in zip(result["labels"], result["scores"]):
    print(f"{label}: {score:.4f}")
```

```python
messages = [
    "There is a water leak in Building C and water is spreading near electrical equipment.",
    "The accessible entrance near the main building has been blocked for two days.",
    "I am concerned that one of our suppliers may not meet our sustainability policy."
]

for i, message in enumerate(messages, start=1):
    print(f"\n--- TEST {i} ---")
    print("Message:", message)

    result = classifier(message, labels)

    for label, score in zip(result["labels"], result["scores"]):
        print(f"{label}: {score:.4f}")
```
